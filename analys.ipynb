```python
import pandas as pd

# Đọc file Excel
file_path = "Phân lớp chuyên sâu GĐ2_AI Thực chiến Khóa 1.xlsx"  # thay bằng đường dẫn file của bạn
df = pd.read_excel(file_path)

# Xem tên cột để chắc chắn đúng
print(df.columns)

# Chuẩn hóa tên cột nếu có khoảng trắng dư
df.columns = df.columns.str.strip()

# Thống kê số lượng theo Track chuyên sâu
track_count = df['Track chuyên sâu'].value_counts().reset_index()
track_count.columns = ['Track chuyên sâu', 'Số lượng']

print(track_count)

# (Tuỳ chọn) Thống kê theo Track + Phòng học
track_room = df.groupby(['Track chuyên sâu', 'Phòng học Giai đoạn 2']).size().reset_index(name='Số lượng')

print(track_room)

# (Tuỳ chọn) Xuất ra file Excel mới
with pd.ExcelWriter("ket_qua_thong_ke.xlsx") as writer:
    track_count.to_excel(writer, sheet_name="Theo Track", index=False)
    track_room.to_excel(writer, sheet_name="Track + Phòng", index=False)
```
